WRITE PROGRAM TO FIND GREATEST AMONG 3 NUMBERS

```python

a=int(input("enter a value: "))
b=int(input("enter b value: "))
c=int(input("enter c value: "))

if(a>b and a>c):
  print(a,"is greatest")
elif(b>a and b>c):
  print(b,"is greatest")
else:
  print(c,"is greatest")
```

PRINT FIZZ IF A NUMBER IS DIVISIBLE BY 3, PRINT BUZZ IF A NUMBER IS DIVISIBLE BY 5, PRINT FIZZBUZZ IF IT IS DIVISIBLE BY BOTH 3 AND 5

```python
a=int(input("enter a value: "))
if (a%3==0):
  print("FIZZ")
elif(a%5==0):
  print("BUZZ")
elif (a%3==0 and a%5==0):
  print("FIZZBUZZ")
```

PRINT FIZZ IF A NUMBER IS DIVISIBLE BY ONLY 3, PRINT BUZZ IF A NUMBER IS DIVISIBLE BY ONLY 5, PRINT FIZZBUZZ IF IT IS DIVISIBLE BY BOTH 3 AND 5

```python
a=int(input("enter a value: "))
if (a%3==0 and a%5!=0):
  print("FIZZ")
elif(a%5==0 and a%3!=0):
  print("BUZZ")
elif (a%3==0 and a%5==0):
  print("FIZZBUZZ")
else:
  print("nothing")
```

PROGRAM ON "MATCH"

```python
num=int(input("enter any number: "))
match num:
  case 1:
    print("Sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednessday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("default")
```
