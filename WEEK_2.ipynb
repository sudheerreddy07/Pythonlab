#PRINT FIZZ IF A NUMBER IS DIVISIBLE BY 3 AND PRINT BUZZ IF A NUMBER IS DIVISIBLE BY 5 AND PRINT FIZZBUZZ IF A NUMBER IS DIVISIBLE WITH BOTH 3 AND 5

```python
num = int(input())
if(num%3 == 0 and num%5 == 0):
   print("FIZZBUZZ")
elif(num%3 == 0):
   print("FIZZ")
elif(num%5 == 0):
   print ("BUZZ")
else:
   print("this is not divisible by 3 and 5")
```

#PROGRAM ON MATCH


```python
num = int(input("enter the number" ))
match num:
  case 1:
    print("MONDAY")
  case 2:
    print("TUESDAY")
  case 3:
    print("WEDNESDAY")
  case 4:
    print("THURSDAY")
  case 5:
    print("FRIDAY")
  case 6:
    print("SATURDAY")
  case 7:
    print("SUNDAY")
  case _:
    print("INVALID")





```
