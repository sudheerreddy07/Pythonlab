#PROGRAM THAT COUNT THE NO OF VOWELS AND CONSONANTS IN A STRING

```python
a = str(input())
b = a.lower()
vowels = "aeiou"
count_c = 0
count_v = 0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no of vowels:",count_v)
print("no of consonents:",count_c)
```

#PROGRAM THAT CHECKS THE SUB STRING IS THE ORIGINAL STRING IF YES PRINT THE INDEX OF THE SUBSTRING

```python
s1 = input("enter a string :")
s2 = input("enter a sub string :")
if s2 in s1:
  print(s1.find(s2))
else:
  print("not found")
```
