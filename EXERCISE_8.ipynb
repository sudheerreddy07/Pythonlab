**FILE HANDLING**

BASIC OPERATIONS

```python
file=open("data.txt","w")
content="New line 1"
file.write(content)   #write
print("file created")
file.close()
```

```python
file=open("data.txt","a")   #append
file.writelines(["\nNew line 2","\nNew line 3"])
file.close()
```

```python
with open("data.txt","r") as file:
  content=file.read(5)   #read
  print(content)
  print(file.tell())
  file.seek(2)
  print(file.read())
```
