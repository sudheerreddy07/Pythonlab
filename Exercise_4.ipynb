#6.ITERATION

```python
L1 = [1,73,'a',"sudheer"]
L2 = [2,80,'c',"sujan"]

print("CONCENTRATION")
print(L1+L2)
print()
print("INDEXING")
print(L1)
print(L2)
print(L1[2])
print(L1[3])
print(L2[2])
print(L2[3])
print()
print("SLICING")
print(L1[0:2:1])
print(L2[2:0:3])
print()
print("REPETION")
print(L1*3)
print(L2*5)
print()
print("MEMBERSHIP OPERATOR")
print("73 in L1")
print("80 in L2")
print()
```
