#GIVEN NUMBER IS PALINDROME OR NOT

```python
def palindrome(num):
  original = num
  rev = 0
  while(num != 0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  if(rev==original):
    return True
  else:
    return False
print(palindrome(121))
print(palindrome(123))

```

#WRITE A PROGRAM TO REVERSE A STRING USING FUNCTIONS

```python
def reverse_str(s):
  rev = ""
  for i in range(len(s)-1,-1,-1):
    rev = rev + s[i]
  return rev
print(reverse_str("hello"))
```

#FUNCTION THAT RETURNS MULTIPLE VALUES

```python
def multiple_values(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"SUM IS {res_sum},COUNT IS {res_count},THE AVERAGE IS {res_avg}")
```
