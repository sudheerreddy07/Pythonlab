#**TUPLES**


#TUPLE OPERATIONS


```python
t1 = (1,2,3)
t2 = (4,5,6)
print(type(t1))
print(type(t2))
print(t1+t2)
print(t1*3)
for i in t1:
  print(i)
print(len(t1))
if 3 in t1:
  print("YES")
else:
  print("NO")

```

#TUPLE METHODS

```python
t3 = (5,6,7,8,9,10)
print(t3.count(8))
print(t3.index(9))
```

#TUPLE FUNCTIONS

```python
print(sum(t1))
print(max(t2))
print(min(t3))
```

#TUPLE ZIP

```python
print(tuple(zip(t1,t2,t3)))
print(list(zip(t1,t2,t3)))
```

#TUPLE SPLIT

```python
t4 = tuple(map(int,input().split()))
print(t4)
```

#WRITE THE PROGRAM TO COUNT THE NUMBER OF VOWELS IN A STRING(NO CONTROL FLOW ALLOWED)

```python
s = str(input("ENTER A STRING:"))
s1 = s.lower()
vowels_count=s1.count("a")+s1.count("e")+s1.count("i")+s1.count("o")+s1.count("u")
print("THE NO OF VOWELS ARE ",vowels_count)
```

# **SETS**

#SET OPERATIONS


```python
A = {1,2,3,4,5}
B = {5,6,7,8,9}
print(A|B)
print(A&B)
print(A-B)
print(A^B)
```

#SET METHODS

```python
A = {1,2,3,4,5}
A.add(6)
print(A)
A.update([3,4])
print(A)
A.discard(2)
print(A)
A.remove(1)
print(A)
A.pop()
print(A)
A.clear()
print(A)
```
